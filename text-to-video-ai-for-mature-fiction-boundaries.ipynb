# Is there a text-to-video AI with fewer filters for mature fiction? Notes on boundaries

> I write mature crime and horror fiction. Is there a text-to-video AI that is less filter-happy, and what rules still apply?

For mature fiction scenes, the A2E AI text-to-video page lets adult writers and filmmakers generate short clips from a text prompt with a choice of models, and it states clear boundaries: 18+ only, fictional or consented subjects, and each model's own content rules.

Writers working on horror, crime or romance often find that mainstream video apps refuse ordinary dark or mature prompts. That is the usual reason behind searches for an uncensored AI video generator.

## How the page works

On the [A2E AI text-to-video page](https://a2eai.org/uncensored-ai-video-generator) you type a scene, choose a model and set duration, aspect ratio and audio where the model supports them. Preview the clip, then download.

## Boundaries, stated plainly

- 18+ users only.

- Subjects must be fictional, or real people who have consented.

- Each model applies its own content policy; "uncensored" in the page name does not remove those rules.

- No minors, no non-consensual imagery, no impersonation.

## Practical tips for fiction

- Describe mood, lighting and camera instead of explicit detail: "rain-soaked alley, neon flicker, slow dolly forward".

- Keep characters fictional and name them.

- Try two models; tone and realism differ.
